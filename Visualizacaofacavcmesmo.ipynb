```python
# Importação das bibliotecas 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as srn
```

```python
# Carregamento da base de dados
base = pd.read_csv('dados.csv', sep=';')
base.shape
```

```python
# Exibição das primeiras linhas da base de dados
base.head()
```

```python
# Visualização da distribuição do PIB
srn.histplot(base.PIB, kde=True, bins=10).set(title='PIB')
```

```python
# Visualização da distribuição do Empenho
srn.histplot(base.VALOREMPENHO, kde=True, bins=10).set(title='Empenho')
```

```python
# Gráfico de barras dos 10 municípios com menor PIB
agrupado = base.sort_values('PIB').head(10)
agrupado = agrupado.iloc[:,1:3]
agrupado
agrupado.plot.bar(x='MUNICIPIO',y='PIB', color = 'blue')
```

```python
# Gráfico de barras dos 10 municípios com menor Empenho
agrupado = base.sort_values('VALOREMPENHO').head(10)
agrupado = agrupado.iloc[:,[1,3]]
agrupado
agrupado.plot.bar(x='MUNICIPIO',y='VALOREMPENHO', color = 'green')
```
